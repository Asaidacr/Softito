### 🔸 Soru 15:
Özyinelemeli (recursive) bir fonksiyon yazın: `faktoriyel(n)` - n! hesaplasın.
- 0! = 1
- n! = n × (n-1)!

5!, 10! değerlerini hesaplayın.

```python
# Cevabınızı buraya yazın:
def faktoriyel(n):
    if n == 0:
        return 1
    # Özyinelemeli adım
    else:
        return n * faktoriyel(n - 1)

sonuc_5 = faktoriyel(5)
sonuc_10 = faktoriyel(10)

print(f"5! = {sonuc_5}")
print(f"10! = {sonuc_10}")
```

### 🔸 Soru 16:
`lambda` fonksiyonu kullanarak:
- Bir sayının karesini hesaplayan lambda
- İki sayının büyüğünü döndüren lambda
- `sorted()` ile bir isim listesini son harfe göre sıralayan lambda

```python
# Cevabınızı buraya yazın:
isimler = ["Zeynep", "Ali", "Mahmut", "Su", "Berk"]
karesi = lambda x: x**2

buyuk_olan = lambda a, b: a if a > b else b

isimler = ["Zeynep", "Ali", "Mahmut", "Su", "Berk"]
# key=lambda x: x[-1] ifadesi sorted fonksiyonuna "son harfe bakarak sırala" der
son_harfe_gore_sirali = sorted(isimler, key=lambda x: x[-1])

# Sonuçların ekrana yazdırılması
print(f"5'in karesi: {karesi(5)}")
print(f"12 ve 8'den büyük olanı: {buyuk_olan(12, 8)}")
print(f"Son harfe göre sıralı liste: {son_harfe_gore_sirali}")
```

### 🔸 Soru 18 (Bütünleştirici):
Basit bir **öğrenci not sistemi** yazın:

- `ogrenciler` adında boş bir sözlük oluşturun (anahtar: isim, değer: notlar listesi)
- `ogrenci_ekle(isim)`: Yeni öğrenci ekler
- `not_ekle(isim, not_)`: Öğrenciye not ekler. Öğrenci yoksa hata mesajı verir.
- `ortalama_hesapla(isim)`: Öğrencinin not ortalamasını döndürür
- `sinif_raporu()`: Tüm öğrencileri ve ortalamalarını yazdırır, sınıf ortalamasını gösterir

En az 3 öğrenci ve her birine 3 not girerek test edin.

```python
# Cevabınızı buraya yazın:

# 1. Boş sözlük oluşturma
ogrenciler = {}

def ogrenci_ekle(isim):
    """Yeni bir öğrenciyi sisteme boş bir liste ile ekler."""
    if isim not in ogrenciler:
        ogrenciler[isim] = []
        print(f"Sisteme eklendi: {isim}")
    else:
        print(f"Hata: {isim} zaten sistemde kayıtlı.")

def not_ekle(isim, not_):
    """Belirtilen öğrenciye yeni bir not ekler."""
    if isim in ogrenciler:
        ogrenciler[isim].append(not_)
        print(f"{isim} adlı öğrenciye {not_} notu eklendi.")
    else:
        print(f"Hata: '{isim}' isminde bir öğrenci bulunamadı!")

def ortalama_hesapla(isim):
    """Öğrencinin not ortalamasını hesaplar ve döndürür."""
    notlar = ogrenciler.get(isim, [])
    if not notlar:
        return 0
    return sum(notlar) / len(notlar)

def sinif_raporu():
    """Tüm öğrencileri, ortalamalarını ve genel sınıf ortalamasını yazdırır."""
    print("\n--- SINIF RAPORU ---")
    toplam_ortalama = 0

    if not ogrenciler:
        print("Sistemde kayıtlı öğrenci yok.")
        return

    for isim in ogrenciler:
        ort = ortalama_hesapla(isim)
        toplam_ortalama += ort
        print(f"Öğrenci: {isim.ljust(10)} | Notlar: {ogrenciler[isim]} | Ortalama: {ort:.2f}")

    genel_ort = toplam_ortalama / len(ogrenciler)
    print("-" * 20)
    print(f"Sınıf Genel Ortalaması: {genel_ort:.2f}")
    print("-" * 20)

# --- TEST AŞAMASI ---

# 3 Öğrenci ekleyelim
ogrenci_ekle("Ahmet")
ogrenci_ekle("Selin")
ogrenci_ekle("Can")

# Her birine 3'er not girelim
not_ekle("Ahmet", 85)
not_ekle("Ahmet", 90)
not_ekle("Ahmet", 78)

not_ekle("Selin", 92)
not_ekle("Selin", 88)
not_ekle("Selin", 95)

not_ekle("Can", 60)
not_ekle("Can", 75)
not_ekle("Can", 70)

# Raporu görelim
sinif_raporu()
```
